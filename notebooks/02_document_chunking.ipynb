```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = """Text splitting is a crucial step in RAG pipelines. It breaks large documents into smaller chunks. These chunks must fit within model context windows. Overlapping chunks help preserve context between boundaries.
"""
```

```python
spliter = RecursiveCharacterTextSplitter(chunk_size=60,chunk_overlap=15)
```

```python
texts = spliter.split_text(text)
```

```python
for txt in texts:
    print(txt)
```

```python
loader = PyPDFLoader(file_path="../data/Profile.pdf")
spliter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=15)
```

```python
docs = loader.load()
```

```python
chunks = spliter.split_documents(docs)
```

```python
for chunk in chunks:
    print(chunk.metadata)
    print()
    print(chunk.page_content)
    print()
    print()
```
